```python
df.columns=df.columns.str.lower()
```

```python
df=df.rename(columns={'rating':'age_rating'})
```

```python
df.columns
```

```python
df.head()
```

```python
# creat a column age_rating_tier
age_rating_map ={
    '7+':'Kids',
'16+':'Teens',
'18+':'Adults'
}
df['age_rating_tier']=df['age_rating'].map(age_rating_map)
```

```python
df[['age_rating','age_rating_tier']].head(10)
```

```python
df.head(8)
```

```python
df=df.rename(columns = {'year':'release_year'})
```

```python
df.head()
```

```python
# create a column release_era

bins=[0,1989,1999,2009,2019,2030]
labels=['Classic','90s','2000s','2010s','Modern']
df['release_era']=pd.cut(df['release_year'],bins=bins,labels=labels)
```
